```python
# Knowledge Base (Rules in IF-THEN format)
knowledge_base = {
    "flu": [["cough", "fever"]],
    "fever": [["sore_throat"]]
}

# Known facts
facts = {"sore_throat", "cough"}

# Backward Chaining Function
def backward_chaining(goal):

    # Check whether the goal is already a known fact
    if goal in facts:
        return True

    # Check whether the goal exists in the knowledge base
    if goal in knowledge_base:

        # Check each rule associated with the goal
        for conditions in knowledge_base[goal]:

            # Verify whether all conditions are satisfied
            if all(backward_chaining(cond) for cond in conditions):
                return True

    # If no rule or fact supports the goal
    return False


# Query: Does the patient have flu?
query = "flu"

if backward_chaining(query):
    print(f"The patient is diagnosed with {query}.")
else:
    print(f"The patient does NOT have {query}.")
```
