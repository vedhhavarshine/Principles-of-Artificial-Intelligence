```python
# Function to check if two predicates can be unified
def unify(x, y, theta=None):
    if theta is None:
        theta = {}

    if x == y:
        return theta

    if isinstance(x, str) and x.islower():
        if x in theta:
            return unify(theta[x], y, theta)
        theta[x] = y
        return theta

    if isinstance(y, str) and y.islower():
        if y in theta:
            return unify(x, theta[y], theta)
        theta[y] = x
        return theta

    if isinstance(x, list) and isinstance(y, list) and len(x) == len(y):
        for a, b in zip(x, y):
            theta = unify(a, b, theta)
            if theta is None:
                return None
        return theta

    return None


# Resolution function for this simple rule format
def resolution(kb, query, facts):
    for clause in kb:
        premise, conclusion = clause

        # Check whether the query matches the conclusion
        if unify(conclusion, query) is not None:
            # Check whether the premise is a known fact
            for fact in facts:
                if unify(premise, fact) is not None:
                    return True

    return False


# Rule: Human(John) -> Mortal(John)
knowledge_base = [
    [["Human", "John"], ["Mortal", "John"]]
]

# Known facts
facts = [
    ["Human", "John"]
]

# Query
query = ["Mortal", "John"]

# Apply resolution
if resolution(knowledge_base, query, facts):
    print("Query is resolved: John is Mortal")
else:
    print("Query could not be resolved")
```
